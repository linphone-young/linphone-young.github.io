```python
import json
from decimal import Decimal
from numpy import int64
class MyDecoder(json.JSONDecoder):
    def decode(self, s):
        return super().decode(s.replace("\n","").lower())
    
def as_complex(dct):
    if 'complex' in dct:
        dct['complex'] = complex(dct['complex']['real'], dct['complex']['imag'])
        
    return dct

test4 = json.load(open(file="text.json")
                  ,cls=MyDecoder
                  ,object_hook=as_complex
                  ,parse_float=Decimal
                  ,parse_int=int64
                  )
print(test4,type(test4['config']['port']),type(test4['split_ratio']))
```

```python
test5 = json.loads('{"name": "Alice", "age": 30, "is_student": false}'
                  ,cls=MyDecoder
                  ,object_hook=as_complex
                  ,parse_float=Decimal
                  ,parse_int=int64
                  )
print(test5,type(test5['name']),type(test5['age']),type(test5['is_student']))
```
