```python
import MDAnalysis as mda
import numpy as np
```

```python
u = mda.Universe('./CH3OH.pdb')
u.atoms.positions = u.atoms.positions@np.array([[0,0,-1],[0,1,0],[1,0,0]])
```

```python
c = u.select_atoms('type C')
o = u.select_atoms('type O')
hc = u.select_atoms('type H and around 1.1 group c', c=c)
ho = u.select_atoms('type H and around 1.1 group o', o=o)
u.select_atoms('around 1.07 group c', c=c)
```

```python
np.linalg.norm(c.positions-o.positions), np.linalg.norm(hc[:1].positions-c.positions), np.linalg.norm(ho[:1].positions-c.positions)
```
